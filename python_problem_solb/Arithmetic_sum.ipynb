```python

# Arithmetic sum equation
import time

n = 50005000

# 1st equation
sat1 = time.time()

total_1 = 0
for i in range(1, n+1):
    total_1 += i

print("Arithmetic sum: ", total_1)
end1 = time.time()
print("1st equation Time: ", end1 - sat1, "Second \n")


# second equation
sat2 = time.time()

total_2 = (n * (n + 1) / 2)
print("Arithmetic sum: ", total_2)
end2 = time.time()
print("2nd equation Time: ", end2 - end1, "Second")
```
